# q 1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in 
the list.

```python
def sum_even(a):
    s=0
    for i in a:
        if i%2==0 :
            s=s+i    
    return s
```

```python
sum_even([1,2,3,4,5,6,7,8])
```

# q.4 Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def prime_no(x):
    if x<=1:
        return "not a prime no."
    for i in range(2,int(x**.5)+1):
        if x%i==0:
            return "no,it has multiple  divisior"
        else:
            return "yes,its prime no"
    return "not from 1 to 200"
```

```python
prime_no(7)
```

```python
prime_no(6)
```
